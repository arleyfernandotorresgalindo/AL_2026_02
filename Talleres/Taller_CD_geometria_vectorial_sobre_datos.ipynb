## Ejercicio 2 — Producto punto y similitud coseno

El **producto punto** $c_i \cdot c_j$ mezcla las tres categorías, pero depende de la
escala de cada cliente (un cliente que gasta mucho en todo tendrá productos punto
grandes con cualquiera). Para comparar **el patrón de consumo** sin que el tamaño del
gasto total distorsione la comparación, se usa la **similitud coseno**:

$$
\cos\theta = \frac{c_i \cdot c_j}{\lVert c_i \rVert \, \lVert c_j \rVert}
$$

a) Calcula el producto punto y la similitud coseno entre $c_1$ y $c_2$, y entre $c_1$
y $c_3$.

b) Según la similitud coseno, ¿$c_1$ se parece más a $c_2$ o a $c_3$ en su **patrón**
de consumo? ¿Tiene sentido con la tabla de datos?

c) Calcula la similitud coseno de $c_1$ contra los otros cuatro clientes y determina
cuál es el "más parecido" a $c_1$ (esta es la idea básica detrás de un sistema de
recomendación basado en similitud).


```python
def similitud_coseno(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

# Ejercicio 2
```
