# Bases y cambio de coordenadas

En el [Cuaderno9](Cuaderno9.ipynb) estudiamos independencia lineal, generado y dimensión. Aquí nos concentramos en dos preguntas:

1. ¿Qué es exactamente una **base** y por qué es útil?
2. Si describimos un mismo vector con dos bases distintas, ¿cómo pasamos de unas **coordenadas** a las otras?

## 1. ¿Qué es una base?

Sea $V$ un espacio vectorial. Un conjunto ordenado $\mathcal{B}=\{b_1,\dots,b_n\}\subset V$ es una **base** de $V$ si:

1. $\mathcal{B}$ es **linealmente independiente**, y
2. $\mathcal{B}$ **genera** a $V$: $\operatorname{gen}\{b_1,\dots,b_n\}=V$.

Intuitivamente, una base tiene *suficientes* vectores para construir todo $V$ (genera) y *ninguno sobra* (independencia). El número $n$ de vectores no depende de la base elegida: es la **dimensión** de $V$.

**Ejemplos.**

| Espacio | Base canónica | Dimensión |
|---|---|---|
| $\mathbb{R}^n$ | $\{e_1,\dots,e_n\}$ | $n$ |
| $P_2$ (polinomios de grado $\le 2$) | $\{1,\,x,\,x^2\}$ | $3$ |
| $M_{2\times 2}$ | $\{E_{11},E_{12},E_{21},E_{22}\}$ ($E_{ij}$ tiene un $1$ en la posición $ij$ y ceros en el resto) | $4$ |

Un espacio tiene infinitas bases. Por ejemplo, $\{(1,1),(1,-1)\}$ también es base de $\mathbb{R}^2$.

**No son bases de $\mathbb{R}^2$:**
- $\{(1,2)\}$: es LI pero no genera (solo produce una recta).
- $\{(1,0),(0,1),(1,1)\}$: genera, pero no es LI (el tercero sobra).

> **Criterio práctico en $\mathbb{R}^n$.** $n$ vectores de $\mathbb{R}^n$ forman una base si y solo si la matriz que los tiene como columnas es invertible ($\det\neq 0$).

## 2. Coordenadas respecto a una base

**Teorema (unicidad de la representación).** Si $\mathcal{B}=\{b_1,\dots,b_n\}$ es base de $V$, todo $v\in V$ se escribe de **una única** forma como
$$v = c_1 b_1 + c_2 b_2 + \cdots + c_n b_n .$$

*Prueba.* Existe al menos una forma porque $\mathcal{B}$ genera. Si hubiera dos, $\sum c_i b_i=\sum d_i b_i$, restando queda $\sum (c_i-d_i)b_i=0$, y por independencia $c_i=d_i$ para todo $i$. $\blacksquare$

Los escalares $c_i$ se llaman **coordenadas** de $v$ en la base $\mathcal{B}$ y se escriben como vector columna:
$$[v]_{\mathcal{B}}=\begin{pmatrix}c_1\\ \vdots \\ c_n\end{pmatrix}\in\mathbb{R}^n.$$

Así, una base convierte cualquier espacio de dimensión $n$ (polinomios, matrices, ...) en $\mathbb{R}^n$: trabajamos con listas de números.

> El **orden** de la base importa: si reordenamos los $b_i$, se reordenan las coordenadas.

**Ejemplo.** Sea $\mathcal{B}=\{(1,1),(1,-1)\}$ y $v=(3,1)$. Buscamos $c_1,c_2$ con
$$c_1(1,1)+c_2(1,-1)=(3,1)\iff\begin{cases}c_1+c_2=3\\ c_1-c_2=1\end{cases}\Rightarrow c_1=2,\ c_2=1.$$
Luego $[v]_{\mathcal{B}}=\begin{pmatrix}2\\1\end{pmatrix}$, mientras que en la base canónica $[v]_{\mathcal{E}}=\begin{pmatrix}3\\1\end{pmatrix}$. **Es el mismo vector; solo cambió la "regla" con que lo medimos.**

**Ejemplo en $P_2$.** Con $\mathcal{E}=\{1,x,x^2\}$, el polinomio $p(x)=5-x+2x^2$ tiene $[p]_{\mathcal{E}}=(5,-1,2)^T$.

### 2.1 Ejemplo: matrices simétricas $2\times 2$ con tres bases

Sea $S=\{A\in M_{2\times 2}: A^T=A\}$, el subespacio de matrices simétricas. Todo elemento de $S$ tiene la forma
$$\begin{pmatrix}a&b\\b&c\end{pmatrix},$$
así que depende de tres parámetros libres: $\dim S=3$. Como la entrada $(2,1)$ repite a la $(1,2)$, **cada igualdad de matrices en $S$ da solo 3 ecuaciones**.

Buscaremos la representación del mismo vector
$$A=\begin{pmatrix}5&2\\2&-1\end{pmatrix}$$
en tres bases distintas.

---

**Base 1 (canónica).**
$$\mathcal{E}=\left\{E_1=\begin{pmatrix}1&0\\0&0\end{pmatrix},\ E_2=\begin{pmatrix}0&1\\1&0\end{pmatrix},\ E_3=\begin{pmatrix}0&0\\0&1\end{pmatrix}\right\}.$$

Planteamos $xE_1+yE_2+zE_3=A$:
$$\begin{pmatrix}x&y\\y&z\end{pmatrix}=\begin{pmatrix}5&2\\2&-1\end{pmatrix}
\iff
\begin{cases}x=5\\ y=2\\ z=-1\end{cases}
\qquad\Rightarrow\qquad [A]_{\mathcal{E}}=\begin{pmatrix}5\\2\\-1\end{pmatrix}.$$

El sistema es inmediato: en la base canónica las coordenadas se "leen" directamente de las entradas.

---

**Base 2.**
$$\mathcal{B}=\left\{B_1=\begin{pmatrix}1&0\\0&1\end{pmatrix},\ B_2=\begin{pmatrix}1&0\\0&-1\end{pmatrix},\ B_3=\begin{pmatrix}0&1\\1&0\end{pmatrix}\right\}.$$

Planteamos $xB_1+yB_2+zB_3=A$:
$$\begin{pmatrix}x+y&z\\z&x-y\end{pmatrix}=\begin{pmatrix}5&2\\2&-1\end{pmatrix}
\iff
\begin{cases}x+y=5\\ z=2\\ x-y=-1\end{cases}$$

Sumando la primera y la tercera: $2x=4\Rightarrow x=2$; luego $y=5-2=3$, y $z=2$.
$$[A]_{\mathcal{B}}=\begin{pmatrix}2\\3\\2\end{pmatrix}.$$

*Verificación:* $2\begin{pmatrix}1&0\\0&1\end{pmatrix}+3\begin{pmatrix}1&0\\0&-1\end{pmatrix}+2\begin{pmatrix}0&1\\1&0\end{pmatrix}=\begin{pmatrix}5&2\\2&-1\end{pmatrix}$. ✓

*¿Es base?* La matriz de coeficientes del sistema es $\begin{pmatrix}1&1&0\\0&0&1\\1&-1&0\end{pmatrix}$, con determinante $2\neq 0$. Entonces el sistema tiene solución única para **cualquier** lado derecho: $\mathcal{B}$ genera a $S$, y con lado derecho $0$ solo existe la solución trivial, así que es LI.

---

**Base 3.**
$$\mathcal{C}=\left\{C_1=\begin{pmatrix}1&1\\1&0\end{pmatrix},\ C_2=\begin{pmatrix}0&1\\1&1\end{pmatrix},\ C_3=\begin{pmatrix}1&0\\0&1\end{pmatrix}\right\}.$$

Planteamos $xC_1+yC_2+zC_3=A$:
$$\begin{pmatrix}x+z&x+y\\x+y&y+z\end{pmatrix}=\begin{pmatrix}5&2\\2&-1\end{pmatrix}
\iff
\begin{cases}x+z=5\\ x+y=2\\ y+z=-1\end{cases}$$

Resolvemos por eliminación de Gauss:
$$\left(\begin{array}{ccc|c}1&0&1&5\\1&1&0&2\\0&1&1&-1\end{array}\right)
\xrightarrow{F_2-F_1}
\left(\begin{array}{ccc|c}1&0&1&5\\0&1&-1&-3\\0&1&1&-1\end{array}\right)
\xrightarrow{F_3-F_2}
\left(\begin{array}{ccc|c}1&0&1&5\\0&1&-1&-3\\0&0&2&2\end{array}\right).$$

Sustitución hacia atrás: $z=1$, $\ y=-3+z=-2$, $\ x=5-z=4$.
$$[A]_{\mathcal{C}}=\begin{pmatrix}4\\-2\\1\end{pmatrix}.$$

*Verificación:* $4\begin{pmatrix}1&1\\1&0\end{pmatrix}-2\begin{pmatrix}0&1\\1&1\end{pmatrix}+1\begin{pmatrix}1&0\\0&1\end{pmatrix}=\begin{pmatrix}5&2\\2&-1\end{pmatrix}$. ✓

*¿Es base?* La eliminación produjo 3 pivotes (determinante $=2\neq 0$), así que, igual que antes, $\mathcal{C}$ es LI y genera a $S$.

---

**Conclusión.** La **misma** matriz $A$ tiene tres vectores de coordenadas distintos:
$$[A]_{\mathcal{E}}=\begin{pmatrix}5\\2\\-1\end{pmatrix},\qquad [A]_{\mathcal{B}}=\begin{pmatrix}2\\3\\2\end{pmatrix},\qquad [A]_{\mathcal{C}}=\begin{pmatrix}4\\-2\\1\end{pmatrix}.$$
Pasar de uno a otro sin volver a resolver un sistema desde cero es justamente el tema de la siguiente sección.

## 3. Cambio de coordenadas

Sean $\mathcal{B}=\{b_1,\dots,b_n\}$ y $\mathcal{C}=\{c_1,\dots,c_n\}$ dos bases de $V$. Queremos una matriz que transforme $[v]_{\mathcal{B}}$ en $[v]_{\mathcal{C}}$.

Si $[v]_{\mathcal{B}}=(x_1,\dots,x_n)^T$, entonces $v=x_1b_1+\cdots+x_nb_n$. Como tomar coordenadas es lineal,
$$[v]_{\mathcal{C}} = x_1[b_1]_{\mathcal{C}}+\cdots+x_n[b_n]_{\mathcal{C}}
=\underbrace{\Big(\,[b_1]_{\mathcal{C}}\ \cdots\ [b_n]_{\mathcal{C}}\,\Big)}_{P_{\mathcal{C}\leftarrow\mathcal{B}}}\,[v]_{\mathcal{B}} .$$

**Definición.** La **matriz de cambio de base** de $\mathcal{B}$ a $\mathcal{C}$ es
$$P_{\mathcal{C}\leftarrow\mathcal{B}}=\Big(\,[b_1]_{\mathcal{C}}\ \cdots\ [b_n]_{\mathcal{C}}\,\Big),\qquad [v]_{\mathcal{C}}=P_{\mathcal{C}\leftarrow\mathcal{B}}\,[v]_{\mathcal{B}}.$$

Es decir: **sus columnas son los vectores de la base vieja escritos en la base nueva.**

**Propiedades.**
- $P_{\mathcal{C}\leftarrow\mathcal{B}}$ es invertible y $\big(P_{\mathcal{C}\leftarrow\mathcal{B}}\big)^{-1}=P_{\mathcal{B}\leftarrow\mathcal{C}}$.
- Composición: $P_{\mathcal{D}\leftarrow\mathcal{C}}\,P_{\mathcal{C}\leftarrow\mathcal{B}}=P_{\mathcal{D}\leftarrow\mathcal{B}}$.

### 3.1 El caso de $\mathbb{R}^n$

Sea $\mathcal{E}$ la base canónica. Si $B=(b_1\ \cdots\ b_n)$ es la matriz con los vectores de $\mathcal{B}$ como columnas, entonces
$$P_{\mathcal{E}\leftarrow\mathcal{B}}=B,\qquad P_{\mathcal{B}\leftarrow\mathcal{E}}=B^{-1}.$$
Por tanto, $v = B\,[v]_{\mathcal{B}}$ y $[v]_{\mathcal{B}}=B^{-1}v$.

Para dos bases cualesquiera $\mathcal{B}$ y $\mathcal{C}$ (con matrices $B$ y $C$), pasando por la canónica:
$$P_{\mathcal{C}\leftarrow\mathcal{B}} = C^{-1}B .$$

**Método práctico (sin calcular inversas):** reducir por filas la matriz aumentada
$$\big[\,C \mid B\,\big]\ \xrightarrow{\text{Gauss-Jordan}}\ \big[\,I \mid P_{\mathcal{C}\leftarrow\mathcal{B}}\,\big].$$

### 3.2 Ejemplo en $\mathbb{R}^2$

$\mathcal{B}=\{(1,1),(1,-1)\}$, $\quad\mathcal{C}=\{(1,0),(1,2)\}$.

Reducimos $[\,C\mid B\,]$:
$$\left(\begin{array}{cc|cc}1&1&1&1\\0&2&1&-1\end{array}\right)
\xrightarrow{F_2/2}
\left(\begin{array}{cc|cc}1&1&1&1\\0&1&\tfrac12&-\tfrac12\end{array}\right)
\xrightarrow{F_1-F_2}
\left(\begin{array}{cc|cc}1&0&\tfrac12&\tfrac32\\0&1&\tfrac12&-\tfrac12\end{array}\right).$$

Luego
$$P_{\mathcal{C}\leftarrow\mathcal{B}}=\begin{pmatrix}\tfrac12&\tfrac32\\[2pt]\tfrac12&-\tfrac12\end{pmatrix}.$$

Tomemos $v=(3,1)$, con $[v]_{\mathcal{B}}=(2,1)^T$ (sección 2):
$$[v]_{\mathcal{C}}=\begin{pmatrix}\tfrac12&\tfrac32\\[2pt]\tfrac12&-\tfrac12\end{pmatrix}\begin{pmatrix}2\\1\end{pmatrix}=\begin{pmatrix}\tfrac52\\[2pt]\tfrac12\end{pmatrix}.$$

**Verificación:** $\tfrac52(1,0)+\tfrac12(1,2)=(3,1)=v$. ✓

### 3.3 Ejemplo en $P_2$

$\mathcal{E}=\{1,x,x^2\}$, $\quad\mathcal{C}=\{1,\ 1+x,\ 1+x+x^2\}$.

Es fácil escribir $\mathcal{C}$ en la base $\mathcal{E}$, así que construimos primero $P_{\mathcal{E}\leftarrow\mathcal{C}}$ (columnas $=[c_j]_{\mathcal{E}}$):
$$P_{\mathcal{E}\leftarrow\mathcal{C}}=\begin{pmatrix}1&1&1\\0&1&1\\0&0&1\end{pmatrix},
\qquad
P_{\mathcal{C}\leftarrow\mathcal{E}}=\big(P_{\mathcal{E}\leftarrow\mathcal{C}}\big)^{-1}=\begin{pmatrix}1&-1&0\\0&1&-1\\0&0&1\end{pmatrix}.$$

Para $p(x)=3+2x+x^2$, $[p]_{\mathcal{E}}=(3,2,1)^T$ y
$$[p]_{\mathcal{C}}=\begin{pmatrix}1&-1&0\\0&1&-1\\0&0&1\end{pmatrix}\begin{pmatrix}3\\2\\1\end{pmatrix}=\begin{pmatrix}1\\1\\1\end{pmatrix}.$$

**Verificación:** $1\cdot 1+1\cdot(1+x)+1\cdot(1+x+x^2)=3+2x+x^2$. ✓

## 4. Interpretación geométrica

Una base de $\mathbb{R}^2$ define una **cuadrícula**: los ejes están en las direcciones de $b_1$ y $b_2$, y la unidad de medida en cada eje es la longitud del vector correspondiente. Las coordenadas $[v]_{\mathcal{B}}$ dicen *cuántos pasos* de $b_1$ y de $b_2$ hay que dar para llegar a $v$.

Cambiar de base **no mueve el vector**: cambia la cuadrícula con la que lo describimos. La matriz $P_{\mathcal{C}\leftarrow\mathcal{B}}$ es el "diccionario" que traduce las instrucciones de una cuadrícula a la otra.

Esta idea reaparecerá más adelante: elegir una base adecuada (por ejemplo, de vectores propios) puede volver muy simple un problema complicado.

## 5. Resumen

- **Base** = conjunto LI que genera $V$. Su tamaño es $\dim V$.
- Cada $v$ tiene coordenadas **únicas** $[v]_{\mathcal{B}}$ en una base dada.
- $P_{\mathcal{C}\leftarrow\mathcal{B}}=\big([b_1]_{\mathcal{C}}\cdots[b_n]_{\mathcal{C}}\big)$ y $[v]_{\mathcal{C}}=P_{\mathcal{C}\leftarrow\mathcal{B}}[v]_{\mathcal{B}}$.
- $P_{\mathcal{B}\leftarrow\mathcal{C}}=P_{\mathcal{C}\leftarrow\mathcal{B}}^{-1}$.
- En $\mathbb{R}^n$: $P_{\mathcal{C}\leftarrow\mathcal{B}}=C^{-1}B$, que se obtiene reduciendo $[\,C\mid B\,]\to[\,I\mid P\,]$.
